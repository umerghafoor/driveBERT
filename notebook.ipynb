```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Load the 3D skeleton data
data = np.load('result/X3D.npy')  # shape: (C, T, V, 3)
```

```python
data[0]
```

```python
# Use first frame (T=0) for visualization
skeleton = data[:, 0, :, :]  # shape: (C, V, 3)

# In case shape is (2, V, 3), take one channel (usually first is joint coords)
if skeleton.shape[0] == 2:
    skeleton = skeleton[0]

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot joints
x = skeleton[:, 0]
y = skeleton[:, 1]
z = skeleton[:, 2]
ax.scatter(x, y, z, c='red')

# Optionally connect joints (if you know the bone structure)
# Example: connect joint 0 to 1, 1 to 2, etc. (just for demo)
for i in range(len(skeleton) - 1):
    ax.plot([x[i], x[i+1]], [y[i], y[i+1]], [z[i], z[i+1]], c='blue')

ax.set_title('3D Skeleton Visualization')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()
```
